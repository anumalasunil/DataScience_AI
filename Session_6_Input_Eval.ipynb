## Eval

**To convert the int/float values as per resultes**

- Ex:
       - n1=int(input('Enter the N1 : ')) # 100
       - n2=float(input('Enter the N2 : ')) #100.50
       - n1+n2 # Ouput: 200.5

       - n1=int(input('Enter the N1 : ')) # 100
       - n2=float(input('Enter the N2 : ')) #100
       - n1+n2 # Ouput: 200.0  # here we need to expect value is 200 ,Hence we are using Keyword is **Eval**

```python
n1=eval(input('Enter the N1 : '))
n2=eval(input('Enter the N2 : '))
n1+n2
```

```python
n1=eval(input('Enter the N1 : '))
n2=eval(input('Enter the N2 : '))
n1+n2
```

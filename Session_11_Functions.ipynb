**Function with arguments**

- In below function we have total 5 variabels are available
    - n1,n2,n3,avg and avg1
- input variables are n1,n2 and n3
- the values are given by user to feed the logic to called as **input varaibels**
- output varoables are avg and avg1
- output varaibels are formed by input varibles
- do not touch **output varaibles**

```python
def average(n1,n2,n3):
    """
    We are doing Avearge programme    
    """
    # doc string
    try:
        #n1=float(input('Enter the number n1: '))
        #n2=float(input('Enter the number n2: '))
        #n3=float(input('Enter the number n3: '))
        avg=(n1+n2+n3)/3
        avg1=round(avg,2)
        print(f'The average of {n1}, {n2} and {n3} :{avg1}')
       
    except Exception as e:
        print(e)
```

```python
average(n1=100,n2=201,n3=20)
```

```python
**Postions only vs keyword_only**
- positional only means the position is fixed
    - it will not Except the argument name
    - Before slash arguments called as **Postional only** arguments
    - so dont give argument name before slash

- Keyword only means the arguments Excepts to be provide
    - After star all arguments are **Keyword only arguments**
    - We should definitely provide the argument name

- Before slash : postion only NO

- After star : Keyword only YES
```

**Module**

- module is a single python file

- It contains
    - method
    - functions
    - class(oops conecepets)
    - variables

**Package**
  
    - packages is a collections of modules
    - collection of python file

**Analaogy**

    - Assume that module is a one gift
    - package is a bag collection of gifts

**Methods vs Functions**

    - Method and funcion both are defined for some task with some one 
    - But method define undder one class
    - Function is defined outside the class 

**Two errors**

  -Module not found error
      - means it is not installed in our laptop

  -Attribute error
      - We know that module has functions methods and variables
      - that are not available under this module
- Whenever we see attribute error perform dir on module name

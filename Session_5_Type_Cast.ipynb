## Type_Cast

- Convert one datat type to another datat type
      Ex: int to other data type like vice versa



```python
10+20  # code into 30
```

**Integer to other data types**

 - int to float
 - int to string
 - int to boolean 

```python
n1=100
n1
type(n1)
```

```python
float_n1=float(n1)
type(float_n1),float_n1
```

```python
str_n1=str(n1)
type(str_n1),str_n1
```

```python
bool_n1=bool(n1)
type(bool_n1), bool_n1
```

    - converted from int to float,str,bool and string

```python
float(-100), str(-100),bool(-100)
```

```python
float(0),str(0),bool(0)
```

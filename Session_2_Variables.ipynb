- Variables can be **upper case**
    -NUMBER=100

- Variables can be **lower case**
    -number=200

- Variables are **case sensitive**
    -number =300 , NUMBER =400

- Variables can allow **number as suffix**
    -number123 = 500

- Variables can not allow **number as prefix**
    -123number = 600

- Variables can not allow **spl characters**
    -number&one = 700

- Variables can allow only **underscore**
    -number_one=800

- Varibels can not be **keywords**
    -if=900

- Variables can be **inbuilt functions** but not allowed
    -sum=10

**Exception Handling**

    - Exception handling is a mechanisum to handle the errors very efficient way

    - Usually we will write amny number of lines in a python code

    - Python is a sequential process, if we face an error at one particular line then remaining line will not be executed

    - Sometimes the error lines might not be important , it is more often infra related errors

    - it does not harmful for our python code/script
                         
    - in this situation we want to continue the execution

    - Exception handling catch the errors efficient way with out crash or block the program

    - **try block** 

        - original code we will write here

    - **except block**
        
        - error will catch here

        - if error happened python execute the except block code lines

**try-except-finally**
 
    - try == original code
    - except == error cathing
    - finally == default block
        - finally block always run irrespective of try and except 

```python
try:
    print('Good')
    print('Connection established')
except:
    print('Not Good')
    print('Connection not established')
finally:
    print('anyways')
    print('Connection closed')
```
